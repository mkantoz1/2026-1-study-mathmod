```julia
using DifferentialEquations
using Plots

# Начальные условия и коэффициенты
x0, y0 = 20000.0, 9000.0
u0 = [x0, y0]
tspan = (0.0, 5.0)

a, b = 0.4, 0.8
c, h = 0.5, 0.7

P(t) = sin(t) + 1.0
Q(t) = cos(t) + 1.0

function model1!(du, u, p, t)
    du[1] = -a*u[1] - b*u[2] + P(t)
    du[2] = -c*u[1] - h*u[2] + Q(t)
end

prob1 = ODEProblem(model1!, u0, tspan)
sol1 = solve(prob1, saveat=0.01)

plot(sol1, vars=(0, 1), label="Армия X (Регулярные)", xlabel="Время t", ylabel="Численность", title="Модель 1: Регулярные войска")
plot!(sol1, vars=(0, 2), label="Армия Y (Регулярные)")
```

```julia
using DifferentialEquations
using Plots

x0 = 20000.0
y0 = 9000.0
u0 = [x0, y0]
tspan = (0.0, 5.0)

a, b = 0.4, 0.8
c = 0.00005 # Масштабированный коэффициент для партизан
h = 0.7

P(t) = sin(t) + 1.0
Q(t) = cos(t) + 1.0

function model2!(du, u, p, t)
    x, y = u
    du[1] = -a*x - b*y + P(t)
    du[2] = -c*x*y - h*y + Q(t)
end

prob = ODEProblem(model2!, u0, tspan)
sol = solve(prob, saveat=0.01)

plot(sol.t, [u[1] for u in sol.u], label="Армия X (Регулярные)", color=:blue, xlabel="Время t", ylabel="Численность", title="Модель 2: Регулярные vs Партизаны")
plot!(sol.t, [u[2] for u in sol.u], label="Армия Y (Партизаны)", color=:red)
```
