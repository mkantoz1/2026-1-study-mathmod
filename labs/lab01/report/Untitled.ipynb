```julia
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint

# Parametreler (Senin ekran görüntündeki değerler)
k = 14.4
n = 4.7
fi = 3 * np.pi / 4  # Kaçakçının kaçış açısı

# Başlangıç uzaklıkları (İki farklı senaryo)
r0_1 = k / (n + 1)
r0_2 = k / (n - 1)

# Diferansiyel denklem: dr/d(theta) = r / sqrt(n^2 - 1)
def dr_dtheta(r, theta):
    return r / np.sqrt(n**2 - 1)

# Açı değerleri (theta)
theta = np.linspace(0, 2 * np.pi, 1000)
# İkinci durum için farklı bir başlangıç açısı gerekebilir, ödevde genelde 0 veya -pi alınır
theta_1 = np.linspace(0, 2 * np.pi, 1000)
theta_2 = np.linspace(-np.pi, np.pi, 1000) 

# Diferansiyel denklemin çözümü (Sahil güvenlik yörüngesi)
r1 = odeint(dr_dtheta, r0_1, theta_1)
r2 = odeint(dr_dtheta, r0_2, theta_2)

# Kaçakçının yörüngesi (Düz bir çizgi)
t = np.linspace(0, 20, 100)
# Kaçakçı fi açısıyla uzaklaşıyor
r_kacakci = t
theta_kacakci = np.full_like(r_kacakci, fi)

# --- GRAFİK ÇİZİMİ ---
plt.figure(figsize=(12, 6))

# 1. Durum Grafiği
ax1 = plt.subplot(121, projection='polar')
ax1.plot(theta_1, r1, label='Sahil Güvenlik (Durum 1)', color='green')
ax1.plot(theta_kacakci, r_kacakci, label='Kaçakçı', color='red', linestyle='--')
ax1.set_title("Durum 1: Karşılıklı Hareket ile Başlangıç")
ax1.legend()

# 2. Durum Grafiği
ax2 = plt.subplot(122, projection='polar')
ax2.plot(theta_2, r2, label='Sahil Güvenlik (Durum 2)', color='blue')
ax2.plot(theta_kacakci, r_kacakci, label='Kaçakçı', color='red', linestyle='--')
ax2.set_title("Durum 2: Aynı Yönde Hareket ile Başlangıç")
ax2.legend()

plt.tight_layout()
plt.show()
```
