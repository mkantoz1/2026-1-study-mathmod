```julia
# Параметры задачи
t0 = 0
x0 = 5         # Начальное число знающих людей (N0)
N = 500        # Максимальное количество потенциальных клиентов
t = np.linspace(0, 30, 300)

# Функции платной рекламы и сарафанного радио (константы)
def k(t):
    return 0.055

def p(t):
    return 0.0018

# Дифференциальное уравнение
def f(x, t):
    return (k(t) + p(t) * x) * (N - x)

# Решение ОДУ
x = odeint(f, x0, t)

# Построение графиков
plt.figure(figsize=(9, 5))
plt.plot(t, x, label='Знающие о салоне x(t)', color='blue', lw=2)
plt.axhline(y=N, color='red', linestyle='--', label=f'Предел N = {N}')
plt.title('Распространение рекламы о салоне красоты')
plt.xlabel('Время t')
plt.ylabel('Количество людей')
plt.grid(True)
plt.legend()
plt.show()
```
